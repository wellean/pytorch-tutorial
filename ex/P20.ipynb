```python
import torch
from torch import nn
from torch.nn import ReLU, Sigmoid

input = torch.tensor([[1, -0.5],
                     [-1, 3]])


output = torch.reshape(input, (-1, 1, 2, 2))
print(output.shape)

class nonlinear(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.Relu = ReLU()
        self.Sigmoid = Sigmoid()

    def forward(self, input):
        output = self.Sigmoid(input)
        return output
    
model = nonlinear()
output = model(input)
print(output)
```

```python
# 其实只要from torch import nn。不必从torch.nn引入ReLU和Sigmoid。nn模块中已经包含了这些函数，可以直接使用nn.ReLU()和nn.Sigmoid()来创建实例。

import torch
import torch.nn as nn

class nonlinear(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.Relu = nn.ReLU()   # 这里的nn.ReLU()和nn.Sigmoid()是从torch.nn模块中引入的函数，可以直接使用。
        self.Sigmoid = nn.Sigmoid()

    def forward(self, input):
        output = self.Sigmoid(input)
        return output




# 同理只要import torch就可以了。torch模块中已经包含了nn模块，可以直接使用torch.nn.Module来定义模型类.


```
