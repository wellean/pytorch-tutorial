```python
class Person:
    def __call__(self, name):
        print("__call__被调用了" + name)

    def hello(self, name):
        print("hello被调用了" + name)


person = Person()
person("张三")
person.hello("李四")

# 可以发现，__call__方法是一个特殊的方法，当我们调用一个对象时，Python会自动调用这个对象的__call__方法。
# hello方法是一个普通的方法，需要我们显式地调用它。
```
