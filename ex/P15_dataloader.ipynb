```python
import torchvision

from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter

test_data = torchvision.datasets.CIFAR10(
    root="./dataset",train=True,download=True,transform=torchvision.transforms.ToTensor())

test_loader = DataLoader(test_data, batch_size=64, shuffle=False, num_workers=0, drop_last=False)    

#测试数据集的第一条数据
img, target = test_data[0]
print(img.shape)
print(target)

#以64为batch_size加载数据集的数据

writer = SummaryWriter(log_dir="./logs")


for epocho in range(2):
    step = 0
    for data in test_loader:
        imgs, targets = data
        # print(imgs.shape)
        # print(targets)
        writer.add_images("Epoch_{}".format(epocho), imgs, step)
        step += 1

writer.close()


```
