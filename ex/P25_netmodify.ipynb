```python
import torchvision
import os
# import torchvision.transforms as transforms

# 设置环境变量 TORCH_HOME 来更改预训练模型的下载路径
# 比如这里下载到当前目录下的 models 文件夹中
os.environ['TORCH_HOME'] = './models'

# imagenet数据集只能手动下载，无法自动下载，不能使用torchvision.datasets.ImageNet来加载数据集。
# imagenet有大约150GB的数据，下载和解压都需要很长时间，因此放弃使用imagenet数据集。

# 解决警告：使用 weights 替代已弃用的 pretrained 参数
# VGG16_Weights.DEFAULT 表示使用最新的权重 (等同于 IMAGENET1K_V1)
vgg_16_true = torchvision.models.vgg16(weights=torchvision.models.VGG16_Weights.DEFAULT)

# weights=None 相当于 pretrained=False，不加载预训练权重
vgg_16_false = torchvision.models.vgg16(weights=None)


# CIFAR10数据集可以自动下载且数据量大小适中，因此使用CIFAR10数据集来加载数据集。
train_data = torchvision.datasets.CIFAR10(root="./dataset", train=True, download=True,
                                        transform=torchvision.transforms.ToTensor())

print(vgg_16_true)
```

```python
import torch

vgg_16_true.add_module("add_linear", torch.nn.Linear(1000, 10)  )

print(vgg_16_true)
```

```python
vgg_16_true.classifier.add_module("add_linear", torch.nn.Linear(1000, 10)  )
print(vgg_16_true)

# 加上classifier后，add_linear就会被加到classifier中，而不是vgg_16_true的最后面了。
```

```python
print(vgg_16_false)
vgg_16_false.classifier[6] = torch.nn.Linear(4096, 10)
print(vgg_16_false)
# 如此可以直接修改classifier中的classifier的索引6层即第七层即最后一层全连接层的输出维度为10，适应CIFAR10数据集的10个类别。
```
