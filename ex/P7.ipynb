```python
from torch.utils.data import Dataset
from PIL import Image
import os

class MyDataset(Dataset):

    def __init__(self, root_dir, label_dir):
        self.root_dir = root_dir
        self.label_dir = label_dir
        self.path = os.path.join(root_dir, label_dir)  ## os.path.join()函数用于将多个路径组合后返回
        self.img_list = os.listdir(self.path)        ## self代表类的实例对象，可以通过它来访问类中的属性和方法。把变量赋给对应的self.属性名，就可以在类的其他方法中通过self.属性名来访问这个变量了。

    def __getitem__(self, idx):
        img_name = self.img_list[idx]
        img_item_path = os.path.join(self.path, img_name)
        img = Image.open(img_item_path)
        label = self.label_dir
        return img, label
    
    def __len__(self):
        
        return len(self.img_list)
    
```

练手数据集/test_producelabels  生成图片对应标签

```python
import os

root_dir = '../练手数据集/test_producelabels/train'
target_dir = 'ants_image'
img_path = os.listdir(os.path.join(root_dir, target_dir))
label = target_dir.split('_')[0]
out_dir = 'ants_label'

# 先创建输出目录，避免写文件时报错
out_path = os.path.join(root_dir, out_dir)
os.makedirs(out_path, exist_ok=True)
for i in img_path:
    file_name = os.path.splitext(i)[0]
    with open(os.path.join(out_path, f"{file_name}.txt"), 'w', encoding='utf-8') as f:
        f.write(label)
```
