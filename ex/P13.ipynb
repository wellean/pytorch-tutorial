```python
from PIL import Image
from torch.utils.tensorboard import SummaryWriter

img_path = "E:/Users/15807/Documents/OneDrive - mail.dhu.edu.cn/学习资料/研究生资料/日常资料/DeepLearning/study/PYTORCH-TUTORIAL/data/test_producelabels/train/ants_image/0013035.jpg"

writer = SummaryWriter(log_dir="logs")

img = Image.open(img_path)

from torchvision import transforms

#ToTensor
trans_totensor = transforms.ToTensor()
img_tensor = trans_totensor(img)
writer.add_image("test", img_tensor)

#Normalize
print(img_tensor[0][0][0])
trans_norm = transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
img_norm = trans_norm(img_tensor)
print(img_norm[0][0][0])
writer.add_image("test_norm", img_norm)

#Resize
print(img.size)
trans_resize = transforms.Resize((512, 512))
#img PIL -> resize -> img_resize PIL
img_resize = trans_resize(img)
print(img_resize.size)

#img_resize PIL -> tensor -> img_resize_tensor tensor
img_resize_tensor = trans_totensor(img_resize)
writer.add_image("test_resize", img_resize_tensor, 0)

#resize2
trans_resize2 = transforms.Resize(128)   
#Compose
trans_compose = transforms.Compose([trans_resize2, trans_totensor])
#resize + totensor -> compose
img_compose = trans_compose(img)
print(img_compose.shape)
writer.add_image("test_resize", img_compose, 1)

#RandomCrop
trans_randomcrop = transforms.RandomCrop(128)
trans_compose2 = transforms.Compose([trans_randomcrop, trans_totensor])
img_compose2 = trans_compose2(img)
writer.add_image("test_resize", img_compose2, 2)

#Random2
for i in range(10):
    img_crop = trans_compose2(img)
    writer.add_image("test_crop", img_crop, i)


writer.close()
```
