```python
from torch.utils.tensorboard import SummaryWriter
import numpy as np
from PIL import Image

image_path = "../data/test_producelabels/train/ants_image/0013035.jpg"
img_PIL = Image.open(image_path)
img_array = np.array(img_PIL)   #print(img_array.shape)  # (512, 768, 3)
print(img_array.shape)
writer = SummaryWriter("logs")





writer.add_image("image", img_array, 1, dataformats="HWC")

for i in range(100):
    writer.add_scalar("y=x^2", i**2, i)

writer.close()
```

```python
from torch.utils.tensorboard import SummaryWriter
import numpy as np
from PIL import Image

image_path = "../data/test_producelabels/train/ants_image/5650366_e22b7e1065.jpg"
img_PIL = Image.open(image_path)
img_array = np.array(img_PIL)   #print(img_array.shape)  # (512, 768, 3)
print(img_array.shape)
writer = SummaryWriter("logs")





writer.add_image("image", img_array, 2, dataformats="HWC")

for i in range(100):
    writer.add_scalar("y=x^2", i**2, i)

writer.close()
```

TensorBoard里会显示新的图片，并且step从1变为2。step1还是会保留的，TensorBoard会显示两张图片，分别对应step1和step2。
step的作用是区分不同的训练步骤或者不同的实验设置，可以帮助我们更好地组织和比较不同的结果。

```python
from torch.utils.tensorboard import SummaryWriter
import numpy as np
from PIL import Image

image_path = "../data/test_producelabels/train\\ants_image\\Nepenthes_rafflesiana_ant.jpg"
img_PIL = Image.open(image_path)
img_array = np.array(img_PIL)   #print(img_array.shape)  # (512, 768, 3)
print(img_array.shape)
writer = SummaryWriter("logs")





writer.add_image("test", img_array, 1, dataformats="HWC")

for i in range(100):
    writer.add_scalar("y=x^2", i**2, i)

writer.close()
```

更换了title后，之前的图片对应的title是"image"，现在的图片对应的title是"test"，所以它们会被区分开来，分别显示在TensorBoard里。每个title下的图片都会按照step进行组织和展示。
